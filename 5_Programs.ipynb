1.Write a Python program to demonstrate how to Draw a Bar Plot using Mat plot lib.

```python
import matplotlib.pyplot as plt

#Giving the data to plot
courses = ['Python','Java','C++','Javascript','AI']
students = [50,40,30,60,45]

#Ploting bar graph
plt.bar(courses,students)

#Specifying the title and labels
plt.title('Numbers of students enrolled in each course')
plt.xlabel('courses')
plt.ylabel('Number of students')

#Shows the bar graph
plt.show()
```

2.Write aPython program to demonstrate how to Draw a Scatter Plot using Mat plot lib

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50]
y = [5,15,25,10,30]

plt.scatter(x,y)
plt.title("Sample scatter plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```

3.Write a Python program to demonstrate how to Draw a Histogram Plot using Mat plot lib.

```python
import matplotlib.pyplot as plt

ages = (12,23,45,22,36,27,29,40,50,30,28,33,31,18,21,25)

plt.hist(ages,bins=5)
plt.title("Age Distribution Histogram")
plt.xlabel("Age Range")
plt.ylabel("Frequency")
plt.show()
```

4.Write a Python program to demonstrate how to Draw a Pie Chart using Mat plot lib.

```python
import matplotlib.pyplot as plt

courses = ['Python','Java','C++','AI']
students = [40,25,20,15]

plt.pie(students, labels=courses, autopct='%1.1f%%')
plt.title("Students Enrollment in Different Courses")
plt.show()
```

5.Write a Python program to illustrate Linear Plotting using Matplotlib.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)

plt.title("Simple Linear Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```
